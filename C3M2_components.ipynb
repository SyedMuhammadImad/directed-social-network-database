# Directed social-network project

Completed local database example. Published sample records are fictional.


```python
from pathlib import Path
import sys
ROOT=next(p for p in [Path.cwd(),*Path.cwd().parents] if (p/'social_database.py').exists())
sys.path.insert(0,str(ROOT))
from social_database import *
```

## Import

Create an in-memory database and atomically import the fictional fixture. The original reference CSV is not embedded.


```python
session,Club,Person,friendships,club_members=create_database()
load_data_from_csv(session,Club,Person,friendships,club_members,ROOT/'example_members.csv')
assert session.query(Person).count()==3
assert session.query(Club).count()==2
```

## Directed friendships and club membership

Outgoing friendships differ from incoming friendships; a connection is not automatically mutual.


```python
assert [p.name for p in get_friends_of_person(session,'Alex Example')]==['Blair Example']
assert [p.name for p in get_persons_who_consider_them_friend(session,'Alex Example')]==['Casey Example']
assert [p.id for p in get_club_members(session,'Chess Club')]==[1,2]
assert get_club_members(session,'Missing Club')==[]
```

## Verification

Repeat imports replace the same graph without accumulating duplicates. Full boundary and rollback checks are in the independent tests, not the unavailable platform grader.


```python
load_data_from_csv(session,Club,Person,friendships,club_members,ROOT/'example_members.csv')
assert session.query(Person).count()==3
engine=session.get_bind()
session.close()
engine.dispose()
```
